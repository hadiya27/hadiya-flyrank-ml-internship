## 3. Verify it with queries (grain, counts, missing values, windows)

*I will verify that each row represents one content item for one client on one reporting date. I will use March 2026 as the mid-panel month and check the row count, date window, and data availability.*

```python
from google.colab import userdata

HF_TOKEN = userdata.get("HF_TOKEN")

print("HF_TOKEN loaded:", HF_TOKEN is not None)
```

```python
from huggingface_hub import HfApi

api = HfApi(token=HF_TOKEN)

files = api.list_repo_files(
    repo_id="FlyRank/internship-warehouse",
    repo_type="dataset"
)

print("\n".join(files))
```

```python
# Section 3 — Query 1: verify the grain

from huggingface_hub import hf_hub_download
import pandas as pd

march_file = hf_hub_download(
    repo_id="FlyRank/internship-warehouse",
    filename="fact_content_daily_performance/month=2026-03/data_0.parquet",
    repo_type="dataset",
    token=HF_TOKEN
)

march_df = pd.read_parquet(march_file)

print("Rows:", len(march_df))
print("Columns:", list(march_df.columns))
print("\nFirst 5 rows:")
display(march_df.head())
```

```python
# Section 3 — Query 1: verify the grain

key_cols = ["report_date", "client_hash_id", "content_hash_id"]

total_rows = len(march_df)
unique_keys = march_df[key_cols].drop_duplicates().shape[0]
duplicate_rows = total_rows - unique_keys

print("Total rows:", total_rows)
print("Unique date + client + content combinations:", unique_keys)
print("Duplicate rows:", duplicate_rows)
```

```python
# Section 3 — Query 2: row count and date span

print("Row count:", len(march_df))
print("Minimum report date:", march_df["report_date"].min())
print("Maximum report date:", march_df["report_date"].max())
print("Number of unique report dates:", march_df["report_date"].nunique())
```

```python
# Section 3 — Query 3: availability using IS TRUE

gsc_available = march_df[march_df["gsc_data_available"].eq(True)]
ga4_available = march_df[march_df["ga4_data_available"].eq(True)]

print("Rows where gsc_data_available IS TRUE:", len(gsc_available))
print("Rows where ga4_data_available IS TRUE:", len(ga4_available))
```

## 4. Data limits

*This slice cannot tell us the full performance of every content item because data availability differs across clients and sources. Some rows have GSC data while others do not have GA4 data, so comparisons can be incomplete. The March 2026 slice covers only one month, so it cannot establish long-term trends or causal effects.*

```python
# Section 4 — check a key data limitation

print("Total March rows:", len(march_df))
print("GSC available rows:", march_df["gsc_data_available"].eq(True).sum())
print("GA4 available rows:", march_df["ga4_data_available"].eq(True).sum())

print("\nGSC availability rate:",
      round(march_df["gsc_data_available"].eq(True).mean(), 3))

print("GA4 availability rate:",
      round(march_df["ga4_data_available"].eq(True).mean(), 3))
```
